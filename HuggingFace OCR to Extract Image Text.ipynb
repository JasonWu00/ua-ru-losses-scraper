```python
from paddleocr import PaddleOCR

# Based on code from this Hugging Face space:
# https://huggingface.co/spaces/pragnakalp/OCR-image-to-text/blob/main/app.py

# declare the OCR outside of func; we do not need to redeclare it every time we ocr
ocr = PaddleOCR(lang='en', use_textline_orientation=True,
                use_doc_unwarping=False,
                use_doc_orientation_classify=False,
                device='gpu',
                text_recognition_model_name='PP-OCRv5_server_rec')
```

```python
def ocr_with_paddle(img: str) -> str:
  """
  Uses PaddleOCR to look for dates in an image.

  ## Parameters
  img: a link to an image on the Internet.

  ## Output
  one or more dates found in the image, separated by |.
  """
  result = ocr.predict(img)
  output = ""
  for thing in dict(result[0])["rec_texts"]: # see docstring for context
    if thing != "":
      if output != "":
        output += "|"
      output += thing
  if output == "":
    return None
  print(f"OCR returns: {output}")
  return output
```

```python
UNSCANNED_LINK_KEYWORD = "UNSCANNED"
```

```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
import time

TIMEOUT = 60

def ocr_dates_apply(inp):
  directproof = ""
  fulldate = ""
  start_time = time.perf_counter()
  #print(inp)
  if "postimg" not in inp["proof"]: # I am not bothering with twitter (or any other) links
    fulldate = None
    directproof = inp["proof"]
    print("Returning non-postimg link")
    return pd.Series([directproof, fulldate])
  # I need to save the direct links anyways
  #if "directproof" in inp.columns and 
  if pd.notna(inp["directproof"]):
    # we ran this code on this row before; skip
    print("Skipping an already-entered row")
    return pd.Series([inp["directproof"], inp["fulldate"]])
  try:
      r = requests.get(inp["proof"], timeout=TIMEOUT)
      soup = BeautifulSoup(r.content, 'html.parser')
      #print(soup)
      direct_img_link = soup.find("meta", attrs={"property": "og:image"}).get("content")
      directproof = direct_img_link
      end_time = time.perf_counter()
  except: # on the off chance that the webside loads nothing
      print("Requests timed out; returning None, None")
      return pd.Series([None, None])
  print(f"Grabbing the full link at {direct_img_link} took {end_time - start_time} seconds to run")
  if pd.notnull(inp["day"]): # do not disturb any existing dates
    fulldate = "" + str(int(inp["day"])) + "/" + str(int(inp["month"])) + "/" + str(int(inp["year"])+2000)
    return pd.Series([directproof, fulldate])
  else: # we need to try ocr
    start_time = time.perf_counter()
    try:
        fulldate = ocr_with_paddle(direct_img_link) # clean up multi-date-stacks later manually
        end_time = time.perf_counter()
        print(f"Running OCR on {direct_img_link} took {end_time - start_time} seconds to run")
        return pd.Series([directproof, fulldate])
    except:
        print("OCR failed for some reason (probably a timeout); returning link and unscanned link keyword")
        return pd.Series([directproof, UNSCANNED_LINK_KEYWORD])
# print(len(ru_losses)) # 19955; evidence where the proof is not twitter
# print(ru_losses[["day", "month", "year"]][:20])
# print(len(ru_losses[pd.notna(ru_losses["year"])])) # 13262; evidence where the ddmmyy is in the webpage and not img
```
